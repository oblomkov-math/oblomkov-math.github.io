# In this file we work on the example of the intesection of Schubert cell that is not an affine space

The semigroup is $\Gamma=<6,9,22>$ and
the semimodule $\Delta$ has generators $0,3,11,14.$
In more details: d=3 and we have 3-cabling of <2,3>,

$x=t^6, y=t^9+\lambda t^{13}$

$\Delta=\Delta_0 \cup \Delta_1 \cup \Delta_2$

$\Delta_0=3 \mathbb{Z}_{\ge 0},\quad \Delta_1=22+3 \mathbb{Z}_{\ge 0},\quad \Delta_2=11+ 3 \mathbb{Z}_{\ge 0}$

To match with our usual notations we set $d=3$, and $n=2,m=3$, $s=4$

Here we define the base ring for our computations

```python
RR.<t,ll,g01,g02,g04,g05,g07,g08,g010,g013,g016,g019,g31,g32,g34,g35,g37,g310,g313,g316,g112,g115,g118,g142,g145> = PolynomialRing(QQ,25)
```

Semi-module $\Delta_j$ has $2$-generators $a_{ji}$. Which are in our case
$$ a_{00}=0, a_{01}=3  \mbox{ for } \Delta_0$$
$$ a_{10}=22, a_{10}=25 \mbox{ for } \Delta_1$$
$$ a_{20}=11, a_{21}=14 \mbox{ for }\Delta_2$$

The corresponding generators are 
$$ g_a=t^a+\sum_{k\in Gaps(\Delta)_{>a}} g_{a;k-a}t^k $$

```python
g0 = 1+g01*t+g02*t^2+g04*t^4+g05*t^5+g07*t^7+g08*t^8+g010*t^10+g013*t^13+g016*t^16+g019*t^19
g3 = t^3+g31*t^4+g32*t^5+g34*t^7+g35*t^8+g37*t^10+g310*t^13+g313*t^16+g316*t^19

g11 = t^11+g112*t^13+g115*t^16+g118*t^19
g14 = t^14+g142*t^16+g145*t^19

g22 = t^22
g25 = t^25
```

The ring generators

```python
x = t^6
y = t^9+ll*t^13
```

Syzygy relation $g0*y-g3*x$ has degree 9.
Below is the computation of syzygy 
$$s_{9;9}=s_{9;\infty}$$

```python
s90 = g0*y-g3*x
k92 = s90.coefficient(t^11)
s92 = s90-k92*g11  # elimination of t^11
k93 = s92.coefficient(t^12)
s93 = s92-k93*x^6*g0  # elimintion of t^12
k95 = s93.coefficient(t^14)
s95 = s93-k95*g14  # elimination of t^14
k96 = s95.coefficient(t^15)
s96 = s95-k96*g3*x^2  # elimination of t^15
k98 = s96.coefficient(t^17)
s98 = s96-k98*g11*x # elimination of t^17
k99 = s98.coefficient(t^18)
s99 = s98-k99*x^3*g0 # elimination of t^18
EqS9= [s99.coefficient(t^10),s99.coefficient(t^13),s99.coefficient(t^16),s99.coefficient(t^19)] # Equations for Jac
```

Syzygy relation $g3*y-g0*x^2$ has degree 12.
Below is the computation of syzygy 
$$s_{12;6}=s_{12;\infty}$$

```python
s120 = g3*y-g0*x^2
k122 = s120.coefficient(t^14)
s122 = s120-k122*g14 #elimination of t^14
## coeff of t^15 in s122 is 0
k125 = s122.coefficient(t^17)
s125 = s122-k125*g11*x # elimination of t^17
k126 = s125.coefficient(t^18)
s126 = s125-k126*g0*x^3 # elimination of t^18
EqS12 = [s126.coefficient(t^13),s126.coefficient(t^16),s126.coefficient(t^19)]
```

```python
EqS9, EqS12
```

We see that the first two equations coinside and the second equations have the same g linear terms which cancel in the sum

```python
EqS9[1]+EqS12[1]
```

The terms that do not depend on $\lambda$ can be factored

```python
(EqS9[1]+EqS12[1]-2*ll).factor()
```

We see that if we set $g02\ne g32$ then we can express $g112-g14$ in terms of $g02$ and $g32$.
Also notice that we impose constraint $g01=g31$ then the leading coefficient of $s90$ is 
$$ (g02-g32)t^{11}+\dots$$

Thus we exclude the generator $g_{11}$ from our generating system thus potentially break a cycle of syzygys in 
$\Delta_2$. 

```python
for i in range(8,15): 
    print(i,s90.coefficient(t^i))
```

But we still have equations EqS9[2] and EqS12[2] and these equations have the same $g$-linear parts

```python
Dif2 = EqS9[2]+EqS12[2]
Dif2Free = Dif2 -ll*Dif2.coefficient(ll)
print(Dif2)
print(Dif2.coefficient(ll).subs(g01 = g31).factor())
```

It is not obvious how one would go about solving this equation, it has no linear terms. However, if we consider constraints for the union $J_\Delta\cup J_{\Delta'}$, $\Delta'=\Delta\cup \{ 19\}$ then equations $EqS12[2]$ and 
$EqS9[3]$ can be dropped.

Indeed, let us consider variety
$$ J_{\Delta; 19}, $$ 
which is defined by the equations $EqS12[i],EqS9[j]$, $i=0,1$, $j=0,1,2$. The functions $EqS12[2],EqS9[3]$ define map
$$ f: J_{\Delta;19}\to \mathbb{C}^2.$$

The fiber $f^{-1}(0,0)$ is $J_\Delta$ and the other fibers are $J_{\Delta'}\times \mathbb{C}$. Also it is clear that 
$$ J_{\Delta;19}=\mathbb{C}^*\times \mathbb{C}^N.$$

Thus to prove that $\chi(J_\Delta)=0$ it is enough to show that $\chi(J_{\Delta'})=0$ but it is easy to see that 
$$ J_{\Delta'}=\mathbb{C}^*\times \mathbb{C}^M$$
